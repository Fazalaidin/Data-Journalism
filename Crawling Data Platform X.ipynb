### Pembersihan Data dan Persiapan Gephi

Langkah-langkah yang akan dilakukan:
1. **Hapus Duplikat**: Menghilangkan tweet yang sama.
2. **Normalisasi User**: Membersihkan spasi pada username.
3. **Ekstraksi Relasi**: Mengambil mention (@user) dan retweet (RT @user) dari teks, serta reply dari metadata.
4. **Agregasi Edge (Weight)**: Menghitung frekuensi interaksi antar user.
5. **Export CSV**: Menyimpan file `Nodes` dan `Edges` dalam format UTF-8.

```python
import re
import numpy as np
import pandas as pd

# 1. Hapus Duplikat (Hanya pada kolom yang bukan list/hashable)
# Kita pilih kolom-kolom utama saja untuk pengecekan duplikat
subset_cols = ['id_str', 'full_text', 'username']
subset_cols = [c for c in subset_cols if c in df.columns]
df = df.drop_duplicates(subset=subset_cols).reset_index(drop=True)

# 2. Inisialisasi Nodes dan Raw Edges
all_users = set()
raw_edges = []

# Fungsi ekstraksi
def extract_mentions(text):
    if isinstance(text, str):
        return [m.strip() for m in re.findall(r'@(\w+)', text) if m.strip()]
    return []

def extract_rt(text):
    if isinstance(text, str):
        match = re.search(r'^RT @(\w+)', text)
        return match.group(1).strip() if match else None
    return None

# 3. Proses Baris per Baris
for _, row in df.iterrows():
    source = str(row['username']).strip()
    if not source or source == 'nan': continue
    all_users.add(source)

    # A. Mentions
    mentions = extract_mentions(row['full_text'])
    for target in mentions:
        if target != source:
            all_users.add(target)
            raw_edges.append({'Source': source, 'Target': target, 'Type': 'Mention'})

    # B. Retweets
    rt_target = extract_rt(row['full_text'])
    if rt_target and rt_target != source:
        all_users.add(rt_target)
        raw_edges.append({'Source': source, 'Target': rt_target, 'Type': 'Retweet'})

    # C. Replies
    reply_target = str(row['in_reply_to_screen_name']).strip()
    if reply_target and reply_target != 'nan' and reply_target != source:
        all_users.add(reply_target)
        raw_edges.append({'Source': source, 'Target': reply_target, 'Type': 'Reply'})

# 4. Buat DataFrame Nodes
nodes_df = pd.DataFrame([{'Id': u, 'Label': u} for u in all_users])

# 5. Agregasi Edges (Weight)
edges_df = pd.DataFrame(raw_edges)
if not edges_df.empty:
    edges_df = edges_df.groupby(['Source', 'Target', 'Type']).size().reset_index(name='Weight')
else:
    edges_df = pd.DataFrame(columns=['Source', 'Target', 'Type', 'Weight'])

print(f"Total Nodes: {len(nodes_df)}")
print(f"Total Edges (Aggregated): {len(edges_df)}")
display(nodes_df.head(), edges_df.head())
```

```python
# 6. Export ke CSV (Final Cleaned Version)
import os
output_path = 'gephi_final'
os.makedirs(output_path, exist_ok=True)

# Simpan ke CSV dengan encoding UTF-8
nodes_df.to_csv(f'{output_path}/nodes.csv', index=False, encoding='utf-8')
edges_df.to_csv(f'{output_path}/edges.csv', index=False, encoding='utf-8')

print(f"File Nodes dan Edges berhasil diperbarui di folder: {output_path}")
print(f"Jumlah Node: {len(nodes_df)}")
print(f"Jumlah Edge: {len(edges_df)}")
print("Silakan unduh file nodes.csv dan edges.csv dari sidebar folder di kiri.")
```
