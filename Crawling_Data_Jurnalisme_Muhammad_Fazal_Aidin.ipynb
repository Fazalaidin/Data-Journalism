```python
import pandas as pd

# Specify the path to your CSV file
file_path = f"tweets-data/{filename}"

# Read the CSV file into a pandas DataFrame
df = pd.read_csv(file_path, delimiter=",")

# Display the DataFrame
display(df)

# Check the length of the DataFrame
num_rows = len(df)
print(f"The number of rows in the DataFrame is: {num_rows}")
```

## Preparing Data for Network Analysis

To perform network analysis in Gephi, we need to create a list of edges, typically represented by 'source' and 'target' nodes. In this Twitter dataset, interactions between users can serve as these edges. I will extract the following interaction types:

1.  **Retweets**: The retweeting user (`username`) is the source, and the original author of the retweeted tweet (`user_rt_screen_name`) is the target.
2.  **Quotes**: The quoting user (`username`) is the source, and the user whose tweet was quoted (`quoted_username`) is the target.
3.  **Replies**: The replying user (`username`) is the source, and the user being replied to (`in_reply_to_screen_name`) is the target.
4.  **Mentions**: The tweeting user (`username`) is the source, and any user mentioned in the `full_text` of the tweet is a target.

```python
import re

# Initialize an empty list to store all interactions
all_interactions = []

# 1. Extract Retweet Interactions
# Check if 'user_rt_screen_name' is present and not all null
if 'user_rt_screen_name' in df.columns:
    retweets = df[df['user_rt_screen_name'].notna()][['username', 'user_rt_screen_name']].copy()
    retweets.columns = ['source', 'target']
    retweets['type'] = 'retweet'
    all_interactions.append(retweets)
    print(f"Extracted {len(retweets)} retweet interactions.")

# 2. Extract Quote Interactions
# Check if 'quoted_username' is present and not all null
if 'quoted_username' in df.columns:
    quotes = df[df['quoted_username'].notna()][['username', 'quoted_username']].copy()
    quotes.columns = ['source', 'target']
    quotes['type'] = 'quote'
    all_interactions.append(quotes)
    print(f"Extracted {len(quotes)} quote interactions.")

# 3. Extract Reply Interactions
# Check if 'in_reply_to_screen_name' is present and not all null
if 'in_reply_to_screen_name' in df.columns:
    replies = df[df['in_reply_to_screen_name'].notna()][['username', 'in_reply_to_screen_name']].copy()
    replies.columns = ['source', 'target']
    replies['type'] = 'reply'
    all_interactions.append(replies)
    print(f"Extracted {len(replies)} reply interactions.")

# 4. Extract Mention Interactions
mentions_list = []
mention_pattern = re.compile(r'@([A-Za-z0-9_]+)')

for index, row in df.iterrows():
    if isinstance(row['full_text'], str):
        mentioned_users = mention_pattern.findall(row['full_text'])
        for mentioned_user in mentioned_users:
            # Avoid self-mentions
            if mentioned_user.lower() != row['username'].lower():
                mentions_list.append({'source': row['username'], 'target': mentioned_user, 'type': 'mention'})

mentions_df = pd.DataFrame(mentions_list)
if not mentions_df.empty:
    all_interactions.append(mentions_df)
    print(f"Extracted {len(mentions_df)} mention interactions.")

# Combine all interactions into a single DataFrame
if all_interactions:
    interactions_df = pd.concat(all_interactions, ignore_index=True)

    # Remove self-loops (where source is the same as target)
    interactions_df = interactions_df[interactions_df['source'].str.lower() != interactions_df['target'].str.lower()]

    # Remove duplicates (e.g., if the same user is retweeted and mentioned)
    interactions_df = interactions_df.drop_duplicates()

    # Display the first few rows of the combined interactions
    print("\nCombined Interactions for Gephi:")
    display(interactions_df.head())
    print(f"Total unique interactions: {len(interactions_df)}")
else:
    print("No interactions found to build the network.")

# You can save this DataFrame to a CSV file for Gephi
# interactions_df.to_csv('gephi_interactions.csv', index=False)
```
